# **3. Numerical Operation**

```r
x <- 5
y <- 3

sum <- x + y
diff <- x - y
mul <- x * y
div <- x / y

print(paste("Sum:", sum))
print(paste("Difference:", diff))
print(paste("Product:", mul))
print(sprintf("Quotient: %.2f", div))


cat("\nSum:", sum, '\n')
cat("Difference:", diff, '\n')
cat("Product:", mul, '\n')
cat("Quotient: %.2f", div, '\n')
```

# **8. Vectors**

```r
grades <- c(90, 88, 96, 83, 91)

grades

cat("\nMean:", mean(grades), "\n")
cat("Max:", max(grades), "\n")
cat("Min:", min(grades), "\n")
cat("Sum:", sum(grades), "\n")
cat("Length:", length(grades), "\n")
```
