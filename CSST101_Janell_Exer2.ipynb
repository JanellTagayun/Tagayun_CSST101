## **4. Using Comparison Operator**

```r
# Consider: (example)
attendance <- 72
grade <- 70

# Determine Whether (True Or False)
attendance < 75
grade < 75
attendance > 75
grade > 75
```

# **5. Using AND and OR**

```r
if (attendance > 75 & grade > 75) {
  print("At Risk")
} else {
  print(" Not At Risk")
}

# OR
if (attendance > 75 | grade > 75) {
  print("At Risk")
} else {
  print(" Not At Risk")
}
```

# **6. Simple IF Statement**

```r
if (grade >= 75) {
  print("Grade => 75, PASS")
} else {
  print("Grade =< 75, FAIL")
}
```

# **7. Multiple Conditions**

```r
if (grade >= 90) {
  print("Excellent")
} else if (grade >= 85) {
  print("Very Good")
} else if (grade >= 80) {
  print("Good")
} else if (grade >= 75) {
  print("Passed")
} else {
  print("Failed")
}
```

# **8. Creating a Function**

```r
class_grade <- function(grade) {
  if (grade >= 90) {
    print("Excellent")
  } else if (grade >= 85) {
    print("Very Good")
  } else if (grade >= 80) {
    print("Good")
  } else if (grade >= 75) {
    print("Passed")
  } else {
    print("Failed")
  }
}

class_grade(100)
```

# **16. Add a Recommendation**

```r
recommend_action <- function(risk_level) {
  if (risk_level == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk_level == "Moderate Risk") {
    return("Academic Monitoring")
  } else {
    return("Continue Regular Monitoring")
  }
}

student$Recommendation <- sapply(student$Risk, recommend_action)

student
```
